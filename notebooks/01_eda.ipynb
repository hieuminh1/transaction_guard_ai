## Interpretation and leakage precautions
Read the measured fraud_rate and class_counts above. High majority-class accuracy alone is misleading. Duplicates count loaded feature/label columns only and are not removed. Missing approved inputs stop validation.

Only step, type, amount and pre-transaction sender balance enter inference. Account identifiers, post-transaction balances, target and simulation flags are excluded. Split on complete time steps. Fit quantile, scaler and encoder on train only. Freeze model/threshold on validation before opening test.

The allowed inputs are a demo availability assumption. Synthetic distributions are not evidence of real bank fraud performance. Average Precision is not trapezoidal PR-AUC.
